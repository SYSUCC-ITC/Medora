```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from math import pi

file_path = r'Fig4\data-e.xlsx'
df = pd.read_excel(file_path)

metrics = ['Accuracy', 'Cited', 'Comprehensible', 'Organized', 
           'Succinct', 'Synthesized', 'Thorough', 'Useful']

stats_data = {}
for metric in metrics:
    col_data = df[metric].dropna()
    mean = np.mean(col_data)
    std = np.std(col_data)
    stats_data[metric] = {'mean': mean, 'std': std, 
                          'mean_minus_std': max(mean - std, 0),
                          'mean_plus_std': mean + std}

categories = metrics
N = len(categories)

angles = [n / float(N) * 2 * pi for n in range(N)]
angles += angles[:1]

fig, ax = plt.subplots(figsize=(12, 12), subplot_kw=dict(projection='polar'))

mean_color = '#2E86AB'
std_color = '#2E86AB'
fill_color = '#2E86AB'
background_color = '#F0F8FF'

theta = np.linspace(0, 2*np.pi, 100)

ax.fill_between(theta, 1, 2, color=background_color, alpha=0.6, zorder=0)
ax.fill_between(theta, 3, 4, color=background_color, alpha=0.6, zorder=0)

mean_values = [stats_data[metric]['mean'] for metric in categories]
mean_values += mean_values[:1]

ax.fill(angles, mean_values, alpha=0.1, color=fill_color, zorder=1)

ax.plot(angles, mean_values, linewidth=2, linestyle='-', 
        color=mean_color, label='Mean', marker='o', markersize=10, 
        markerfacecolor='white', markeredgewidth=2, zorder=2)

for i, metric in enumerate(categories):
    angle = angles[i]
    mean_val = stats_data[metric]['mean']
    lower_val = stats_data[metric]['mean_minus_std']
    upper_val = stats_data[metric]['mean_plus_std']
    
    ax.plot([angle, angle], [lower_val, upper_val], 
            linewidth=3, color=std_color, alpha=0.8, zorder=2)
    
    whisker_width = 0.03
    ax.plot([angle - whisker_width, angle + whisker_width], 
            [lower_val, lower_val], 
            linewidth=3, color=std_color, alpha=0.8, zorder=2)
    ax.plot([angle - whisker_width, angle + whisker_width], 
            [upper_val, upper_val], 
            linewidth=3, color=std_color, alpha=0.8, zorder=2)

plt.rcParams['font.sans-serif'] = ['DejaVu Sans', 'Arial']
plt.rcParams['axes.unicode_minus'] = False

plt.xticks(angles[:-1], categories, size=12, fontweight='bold')
ax.tick_params(axis='x', pad=25)
ax.set_theta_offset(pi/2)
ax.set_theta_direction(-1)

ax.set_ylim(0, 6)

yticks = [1, 2, 3, 4, 5]
ax.set_yticks(yticks)
ax.set_yticklabels([str(tick) for tick in yticks], size=11, color='gray')

ax.grid(True, linestyle='--', linewidth=0.8, alpha=0.7)

ax.set_facecolor('#f8f9fa')
fig.patch.set_facecolor('white')

center_text = f"Mean range: {np.min(mean_values[:-1]):.2f} - {np.max(mean_values[:-1]):.2f}\n"
center_text += f"Overall mean: {np.mean(mean_values[:-1]):.2f}"
plt.text(0, -0.5, center_text, ha='center', va='center', fontsize=10,
         bbox=dict(boxstyle="round,pad=0.5", facecolor="lightgray", alpha=0.5))

plt.tight_layout()

pdf_path = r'\Fig. 4-e.pdf'
plt.savefig(pdf_path, format='pdf', dpi=300, bbox_inches='tight')
print(f"Radar chart saved as PDF: {pdf_path}")

plt.show()

print("=" * 60)
print("Statistics for each indicator:")
print("=" * 60)
for metric in metrics:
    print(f"{metric}:")
    print(f"  Mean: {stats_data[metric]['mean']:.2f}")
    print(f"  Std: {stats_data[metric]['std']:.2f}")
    print(f"  Range: {stats_data[metric]['mean_minus_std']:.2f} - {stats_data[metric]['mean_plus_std']:.2f}")
    print("-" * 40)
```
